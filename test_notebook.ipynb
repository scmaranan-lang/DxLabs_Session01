```python
# choose a random name from a list of names
import random
import time
names = ["Alice", "Bob", "Charlie", "David", "Eve"]
names.append("Emma")
random_name = random.choice(names)

for seconds in range(3, 0, -1):
    print(seconds)
    time.sleep(1)

print(f"Randomly selected name: {random_name}")
```
